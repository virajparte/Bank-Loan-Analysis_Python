```python

import pandas as pd

# Convert dates correctly: DD-MM-YYYY
df['issue_date'] = pd.to_datetime(
    df['issue_date'], format='%d-%m-%Y', errors='coerce'
)

# Find the latest issue date
latest_issue_date = df['issue_date'].max()

# Extract year and month
latest_year = latest_issue_date.year
latest_month = latest_issue_date.month

# Filter applications for the latest month
mtd_date = df[
    (df['issue_date'].dt.year == latest_year) &
    (df['issue_date'].dt.month == latest_month)
]

# Count loan applications
mtd_loan_applications = mtd_date['id'].count()

print(
    f"MTD Loan Applications "
    f"(for {latest_issue_date.strftime('%B %Y')}): "
    f"{mtd_loan_applications:,}"
)
```

Charts

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df['issue_date'] = pd.to_datetime(
    df['issue_date'], dayfirst=True, errors='coerce'
)
```

1. Monthly Trends by Issue Date — Line Chart

```python

monthly = df.groupby(
    df['issue_date'].dt.to_period('M')
).agg(
    applications=('id', 'count'),
    funded_amount=('loan_amount', 'sum'),
    amount_received=('total_payment', 'sum')
).reset_index()

monthly['issue_date'] = monthly['issue_date'].astype(str)

plt.figure(figsize=(12, 5))
plt.plot(monthly['issue_date'], monthly['applications'], marker='o')
plt.title('Monthly Loan Applications')
plt.xlabel('Issue Month')
plt.ylabel('Total Loan Applications')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

2. Regional Analysis by State — Bar Chart

```python

state_data = df.groupby('address_state').agg(
    applications=('id', 'count'),
    funded_amount=('loan_amount', 'sum'),
    amount_received=('total_payment', 'sum')
).sort_values('applications', ascending=False).head(10)

state_data['applications'].sort_values().plot(
    kind='barh', figsize=(10, 6)
)
plt.title('Top 10 States by Loan Applications')
plt.xlabel('Total Loan Applications')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```

3. Loan Term Analysis — Donut Chart

```python

term_data = df.groupby('term')['id'].count()

plt.figure(figsize=(7, 7))
plt.pie(
    term_data,
    labels=term_data.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)
plt.title('Loan Applications by Term')
plt.show()
```

4. Employee Length Analysis — Bar Chart

```python

emp_data = df.groupby('emp_length').agg(
    applications=('id', 'count'),
    funded_amount=('loan_amount', 'sum'),
    amount_received=('total_payment', 'sum')
)

emp_order = [
    '< 1 year', '1 year', '2 years', '3 years', '4 years',
    '5 years', '6 years', '7 years', '8 years', '9 years',
    '10+ years'
]

emp_data = emp_data.reindex(emp_order).dropna()

emp_data['applications'].plot(kind='bar', figsize=(10, 5))
plt.title('Loan Applications by Employment Length')
plt.xlabel('Employment Length')
plt.ylabel('Total Loan Applications')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

5. Loan Purpose Breakdown — Bar Chart

```python

purpose_data = df.groupby('purpose').agg(
    applications=('id', 'count'),
    funded_amount=('loan_amount', 'sum'),
    amount_received=('total_payment', 'sum')
).sort_values('applications')

purpose_data['applications'].plot(
    kind='barh', figsize=(10, 7)
)
plt.title('Loan Applications by Purpose')
plt.xlabel('Total Loan Applications')
plt.ylabel('Loan Purpose')
plt.tight_layout()
plt.show()
```

6. Home Ownership Analysis — Bar Chart

```python

home_data = df.groupby('home_ownership').agg(
    applications=('id', 'count'),
    funded_amount=('loan_amount', 'sum'),
    amount_received=('total_payment', 'sum')
).sort_values('applications', ascending=False)

home_data['applications'].plot(
    kind='bar', figsize=(9, 5)
)
plt.title('Loan Applications by Home Ownership')
plt.xlabel('Home Ownership')
plt.ylabel('Total Loan Applications')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

# Financial Loan Analysis — Key Insights

## 1. Overall Loan Portfolio
- Total loan applications: **38,576**
- Total funded amount: **$435.76M**
- Average interest rate: **12.05%**
- Average debt-to-income (DTI) ratio: Calculate using the `dti` column.

## 2. Monthly Lending Trends
- Loan applications increased from **2,332 in January to 4,314 in December 2021**.
- This represents an increase of approximately **85%**.
- December recorded the highest application volume in the dataset.
- Business insight: Investigate monthly demand patterns to improve lending capacity and resource planning.

## 3. Regional Analysis
- California recorded the highest number of applications: **6,894**.
- New York ranked second with **3,701** applications.
- Florida and Texas followed with **2,773** and **2,664** applications.
- Business insight: Focus on high-volume states while comparing their loan performance and default rates.

## 4. Loan Term Analysis
- 36-month loans: **28,237 applications (73.2%)**.
- 60-month loans: **10,339 applications (26.8%)**.
- The charged-off rate was approximately **10.7% for 36-month loans** and **22.3% for 60-month loans**.
- Business insight: Longer-term loans showed a higher charged-off rate and require further risk assessment.

## 5. Employment Length Analysis
- Borrowers with 10+ years of employment formed the largest group: **8,870 applications**.
- Borrowers with less than one year of employment accounted for **4,575 applications**.
- Business insight: Compare charged-off rates across employment groups to evaluate repayment risk.

## 6. Loan Purpose Analysis
- Debt consolidation was the leading purpose: **18,214 applications (47.2%)**.
- Credit card loans ranked second: **4,998 applications**.
- Small-business loans showed a charged-off rate of approximately **25.6%**, compared with 13.8% overall.
- Business insight: Investigate the risk profile and repayment performance of small-business loans.

## 7. Home Ownership Analysis
- Renters: **18,439 applications**.
- Mortgage holders: **17,198 applications**.
- Homeowners: **2,838 applications**.
- Charged-off rates were approximately 14.6% for renters, 13.0% for mortgage holders, and 14.0% for homeowners.
- Business insight: Consider home ownership alongside income, credit grade, DTI, and repayment history.

## 8. Good Loan vs. Bad Loan Analysis
- Good loans (Fully Paid + Current): **33,243 applications (86.18%)**.
- Bad loans (Charged Off): **5,333 applications (13.82%)**.
- Fully Paid loans accounted for 83.33% of all applications.
- Current loans accounted for 2.85%.
- Business insight: Monitor charged-off loans to improve risk management and collection strategies.

## 9. Key Business Recommendations
1. Monitor monthly application growth and lending demand.
2. Investigate the higher charged-off rate for 60-month loans.
3. Review underwriting performance for small-business loans.
4. Prioritize analysis of debt-consolidation applications because of their high volume.
5. Compare state-level application volumes with default rates and profitability.
6. Track funded amounts, payments received, and loan status together.

## 10. Main Conclusion
The dataset shows growing lending activity, strong demand for debt consolidation, and an overall charged-off rate of approximately 13.82%. Longer-term loans and small-business loans show elevated charged-off rates, highlighting opportunities for further credit-risk analysis and better lending decisions.
